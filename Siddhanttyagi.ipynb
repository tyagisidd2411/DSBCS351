```python
def removeElement(self, nums, val):
        k = 0

        for num in nums:
            if num != val:
                nums[k] = num
                k += 1
```

```python

    def buildArray(self, nums):
        ans = []

        for i in range(len(nums)):
            ans.append(nums[nums[i]])

        return ans
```

```python
def search(self, nums, target):
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        return -1
```
