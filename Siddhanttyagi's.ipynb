```python


%%writefile linkedlist.c

#include <stdio.h>

#define MAX 100

struct Node {
    int data;
    int next;
};

struct Node list[MAX];

int main() {
    int n, i;

    printf("Enter number of nodes: ");
    scanf("%d", &n);

    // Input data
    for (i = 0; i < n; i++) {
        printf("Enter data for node %d: ", i + 1);
        scanf("%d", &list[i].data);

        if (i == n - 1)
            list[i].next = -1;
        else
            list[i].next = i + 1;
    }

    // Display linked list
    printf("\nLinked List: ");

    i = 0;

    while (i != -1) {
        printf("%d", list[i].data);

        if (list[i].next != -1)
            printf(" -> ");

        i = list[i].next;
    }

    printf(" -> NULL\n");

    return 0;
}
```

```python
%%writefile traversal.c

#include <stdio.h>
#include <stdlib.h>

struct Node {
    int data;
    struct Node *next;
};

int main() {
    struct Node *head = NULL;
    struct Node *temp = NULL;
    struct Node *newNode = NULL;

    int n, i, value;

    printf("Enter number of nodes: ");
    scanf("%d", &n);

    // Creating the linked list
    for (i = 0; i < n; i++) {

        newNode = (struct Node *)malloc(sizeof(struct Node));

        printf("Enter data for node %d: ", i + 1);
        scanf("%d", &value);

        newNode->data = value;
        newNode->next = NULL;

        if (head == NULL) {
            head = newNode;
            temp = newNode;
        } else {
            temp->next = newNode;
            temp = newNode;
        }
    }

    // Traversal
    printf("\nLinked List: ");

    temp = head;

    while (temp != NULL) {
        printf("%d", temp->data);

        if (temp->next != NULL) {
            printf(" -> ");
        }

        temp = temp->next;
    }

    printf(" -> NULL\n");

    return 0;
}
```
