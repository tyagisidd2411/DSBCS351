```python


%%writefile linkedlist.c

#include <stdio.h>

#define MAX 100

struct Node {
    int data;
    int next;
};

struct Node list[MAX];

int main() {
    int n, i;

    printf("Enter number of nodes: ");
    scanf("%d", &n);

    // Input data
    for (i = 0; i < n; i++) {
        printf("Enter data for node %d: ", i + 1);
        scanf("%d", &list[i].data);

        if (i == n - 1)
            list[i].next = -1;
        else
            list[i].next = i + 1;
    }

    // Display linked list
    printf("\nLinked List: ");

    i = 0;

    while (i != -1) {
        printf("%d", list[i].data);

        if (list[i].next != -1)
            printf(" -> ");

        i = list[i].next;
    }

    printf(" -> NULL\n");

    return 0;
}
```

```python
%%writefile traversal.c

#include <stdio.h>
#include <stdlib.h>

struct Node {
    int data;
    struct Node *next;
};

int main() {
    struct Node *head = NULL;
    struct Node *temp = NULL;
    struct Node *newNode = NULL;

    int n, i, value;

    printf("Enter number of nodes: ");
    scanf("%d", &n);

    // Creating the linked list
    for (i = 0; i < n; i++) {

        newNode = (struct Node *)malloc(sizeof(struct Node));

        printf("Enter data for node %d: ", i + 1);
        scanf("%d", &value);

        newNode->data = value;
        newNode->next = NULL;

        if (head == NULL) {
            head = newNode;
            temp = newNode;
        } else {
            temp->next = newNode;
            temp = newNode;
        }
    }

    // Traversal
    printf("\nLinked List: ");

    temp = head;

    while (temp != NULL) {
        printf("%d", temp->data);

        if (temp->next != NULL) {
            printf(" -> ");
        }

        temp = temp->next;
    }

    printf(" -> NULL\n");

    return 0;
}
```

```python
%%writefile insertion_kth.c

#include <stdio.h>
#include <stdlib.h>

struct Node {
    int data;
    struct Node *next;
};

int main() {
    struct Node *head = NULL;
    struct Node *newNode = NULL;
    struct Node *temp = NULL;

    int n, i, value, k;

    // Create the linked list
    printf("Enter number of nodes: ");
    scanf("%d", &n);

    for (i = 0; i < n; i++) {

        newNode = (struct Node *)malloc(sizeof(struct Node));

        printf("Enter data for node %d: ", i + 1);
        scanf("%d", &value);

        newNode->data = value;
        newNode->next = NULL;

        if (head == NULL) {
            head = newNode;
        } else {
            temp = head;

            while (temp->next != NULL) {
                temp = temp->next;
            }

            temp->next = newNode;
        }
    }

    // Display original list
    printf("\nOriginal Linked List: ");

    temp = head;

    while (temp != NULL) {
        printf("%d -> ", temp->data);
        temp = temp->next;
    }

    printf("NULL\n");

    // Input position and value
    printf("\nEnter position (K): ");
    scanf("%d", &k);

    printf("Enter data to insert: ");
    scanf("%d", &value);

    // Check valid position
    if (k < 1 || k > n + 1) {
        printf("Invalid position!\n");
        return 0;
    }

    // Create new node
    newNode = (struct Node *)malloc(sizeof(struct Node));
    newNode->data = value;

    // Insertion at beginning
    if (k == 1) {
        newNode->next = head;
        head = newNode;
    }
    else {
        temp = head;

        // Move to node before kth position
        for (i = 1; i < k - 1; i++) {
            temp = temp->next;
        }

        newNode->next = temp->next;
        temp->next = newNode;
    }

    // Display updated list
    printf("\nLinked List after insertion: ");

    temp = head;

    while (temp != NULL) {
        printf("%d -> ", temp->data);
        temp = temp->next;
    }

    printf("NULL\n");

    return 0;
}
```

```python
%%writefile deletion_beginning.c

#include <stdio.h>
#include <stdlib.h>

struct Node {
    int data;
    struct Node *next;
};

int main() {
    struct Node *head = NULL;
    struct Node *newNode = NULL;
    struct Node *temp = NULL;

    int n, i, value;

    // Create the linked list
    printf("Enter number of nodes: ");
    scanf("%d", &n);

    for (i = 0; i < n; i++) {

        newNode = (struct Node *)malloc(sizeof(struct Node));

        printf("Enter data for node %d: ", i + 1);
        scanf("%d", &value);

        newNode->data = value;
        newNode->next = NULL;

        if (head == NULL) {
            head = newNode;
        } else {
            temp = head;

            while (temp->next != NULL) {
                temp = temp->next;
            }

            temp->next = newNode;
        }
    }

    // Display original list
    printf("\nOriginal Linked List: ");

    temp = head;

    while (temp != NULL) {
        printf("%d -> ", temp->data);
        temp = temp->next;
    }

    printf("NULL\n");

    // Deletion from beginning
    if (head == NULL) {
        printf("\nList is empty. Nothing to delete.\n");
    } else {
        temp = head;
        head = head->next;
        free(temp);

        // Display updated list
        printf("\nLinked List after deletion from beginning: ");

        temp = head;

        while (temp != NULL) {
            printf("%d -> ", temp->data);
            temp = temp->next;
        }

        printf("NULL\n");
    }

    return 0;
}
```

```python
%%writefile deletion_end.c

#include <stdio.h>
#include <stdlib.h>

struct Node {
    int data;
    struct Node *next;
};

int main() {
    struct Node *head = NULL;
    struct Node *newNode = NULL;
    struct Node *temp = NULL;
    struct Node *prev = NULL;

    int n, i, value;

    // Create the linked list
    printf("Enter number of nodes: ");
    scanf("%d", &n);

    for (i = 0; i < n; i++) {

        newNode = (struct Node *)malloc(sizeof(struct Node));

        printf("Enter data for node %d: ", i + 1);
        scanf("%d", &value);

        newNode->data = value;
        newNode->next = NULL;

        if (head == NULL) {
            head = newNode;
        } else {
            temp = head;

            while (temp->next != NULL) {
                temp = temp->next;
            }

            temp->next = newNode;
        }
    }

    // Display original list
    printf("\nOriginal Linked List: ");

    temp = head;

    while (temp != NULL) {
        printf("%d -> ", temp->data);
        temp = temp->next;
    }

    printf("NULL\n");

    // Deletion from end
    if (head == NULL) {
        printf("\nList is empty. Nothing to delete.\n");
    }
    else if (head->next == NULL) {
        // Only one node
        free(head);
        head = NULL;
    }
    else {
        temp = head;

        // Move to the last node
        while (temp->next != NULL) {
            prev = temp;
            temp = temp->next;
        }

        // Remove the last node
        prev->next = NULL;
        free(temp);
    }

    // Display updated list
    printf("\nLinked List after deletion from end: ");

    temp = head;

    while (temp != NULL) {
        printf("%d -> ", temp->data);
        temp = temp->next;
    }

    printf("NULL\n");

    return 0;
}
```
